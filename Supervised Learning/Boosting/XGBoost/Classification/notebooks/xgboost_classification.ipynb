# Boosting: XGBoost Classification

This notebook demonstrates XGBoost (Extreme Gradient Boosting) for classification, a high-performance gradient boosting framework.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, roc_curve

np.random.seed(42)

# Note: Install xgboost if not available: pip install xgboost
try:
    import xgboost as xgb
    print(f"XGBoost version: {xgb.__version__}")
except ImportError:
    print("XGBoost not installed. Install with: pip install xgboost")
```
